# 2.1.2 标签质量检查

原题把“低碳行为积极性”当作预测目标。打开数据后先数这个列有几个不同取值。

如果全表只有一个取值，就**不能训练**。不要编造、插补或替换标签，也不要继续做标准化和划分。月生活费列名是金额区间，表里存的是选项代号，不是连续的“元”。

把证据写入 `2.1.2_quality.json`。

```python
import json
from pathlib import Path
import pandas as pd

# 读取原表。这一步只检查，不训练。
df = __________("大学生低碳生活行为的影响因素数据集.xlsx")
target = "低碳行为积极性"
values = pd.to_numeric(df[target], errors="coerce")
print("目标列取值：")
print(values.value_counts(dropna=False))

report = {
    "rows": int(len(df)),
    "columns": int(df.shape[1]),
    "missing_cells": int(df.isna().sum().sum()),
    "duplicate_rows": int(df.duplicated().sum()),
    "target": target,
    "target_nunique": int(values.__________(dropna=True)),
    "target_counts": {str(k): int(v) for k, v in values.value_counts(dropna=False).items()},
    "trainable": bool(values.nunique(dropna=True) >= 2),
}
# 只有一个取值时这里必须为 False。不要改数据来让它通过。
assert report["trainable"] is False
report["decision"] = "目标列只有常量 0，不能训练有区分度的预测模型。不要编造、插补或替换标签。"
Path("2.1.2_quality.json").write_text(json.dumps(report, ensure_ascii=False, indent=2), encoding="utf-8")
print("配套目标列为常量")
print(report["decision"])
```
